โค้ดทดสอบ

```python
from ultralytics import YOLO
import glob
from IPython.display import Image, display

# 1. โหลดโมเดล best.pt ของเรา
model = YOLO('runs/detect/train/weights/best.pt')

# 2. สั่งทำนายภาพ "ทั้งโฟลเดอร์"
# แค่ใส่ source เป็น path ของโฟลเดอร์ โมเดลจะจัดการรูปข้างในให้ทั้งหมด
results = model.predict(source='bottle_detection-7/test/images', save=True)

# 3. โค้ดสำหรับดึงรูปภาพที่ตรวจจับเสร็จแล้วมาแสดงผล (สุ่มมาให้ดู 3 รูปแรก)
latest_predict_folder = sorted(glob.glob('runs/detect/predict*'))[-1]
result_images = glob.glob(f'{latest_predict_folder}/*.jpg')

print(f"ทำนายเสร็จแล้ว! รูปภาพถูกบันทึกไว้ที่: {latest_predict_folder}")
print("ตัวอย่างผลลัพธ์ 3 รูปแรก:")

for img_path in result_images[:3]: # เปลี่ยนตัวเลข 3 เป็นจำนวนที่ต้องการดูได้
    display(Image(filename=img_path, width=600))
```
